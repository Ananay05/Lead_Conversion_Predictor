```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 120

df = pd.read_csv('../data/leads.csv')
df.head()
```

## 1. Basic Info

```python
print('Shape:', df.shape)
print('\nDtype:\n', df.dtypes)
print('\nMissing values:\n', df.isnull().sum())
print('\nConversion rate:', df['converted'].mean().round(4))
```

## 2. Target Distribution

```python
fig, ax = plt.subplots(figsize=(5, 3))
df['converted'].value_counts().plot.bar(ax=ax, color=['#ef233c', '#06d6a0'], edgecolor='white')
ax.set_xticklabels(['Not Converted', 'Converted'], rotation=0)
ax.set_title('Target Class Distribution')
plt.tight_layout()
plt.show()
```

## 3. Numerical Features — Distribution

```python
num_cols = ['age', 'income', 'website_visits', 'time_spent_on_site', 'pages_viewed', 'lead_score', 'follow_up_calls']
df[num_cols].hist(figsize=(12, 8), bins=20, color='#4361ee', edgecolor='white')
plt.suptitle('Numerical Feature Distributions', y=1.02)
plt.tight_layout()
plt.show()
```

## 4. Correlation Heatmap

```python
from sklearn.preprocessing import LabelEncoder

df_enc = df.copy()
for col in ['lead_source', 'industry']:
    df_enc[col] = LabelEncoder().fit_transform(df_enc[col])
df_enc = df_enc.drop(columns=['lead_id'])

plt.figure(figsize=(10, 7))
sns.heatmap(df_enc.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

## 5. Lead Source vs Conversion Rate

```python
source_conv = df.groupby('lead_source')['converted'].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(7, 4))
source_conv.plot.bar(ax=ax, color='#4361ee', edgecolor='white')
ax.set_ylabel('Conversion Rate')
ax.set_title('Conversion Rate by Lead Source')
ax.set_xticklabels(ax.get_xticklabels(), rotation=15)
plt.tight_layout()
plt.show()
```

## 6. Lead Score vs Conversion

```python
fig, ax = plt.subplots(figsize=(7, 4))
for label, color in [(0, '#ef233c'), (1, '#06d6a0')]:
    subset = df[df['converted'] == label]['lead_score']
    ax.hist(subset, bins=20, alpha=0.6, label=['Not Converted', 'Converted'][label], color=color)
ax.set_xlabel('Lead Score')
ax.legend()
ax.set_title('Lead Score Distribution by Outcome')
plt.tight_layout()
plt.show()
```
